# 4.. Exploring the Claude Code Panel in VS Code !

```python
'''




    Here is the chat panel of Claude Code !
    Where u can mention prompt & interact with Claude Code !

	Ask before edit : ( opt for this , bcz in this Claude asks for your approval before making any change to your codebase. ) 
                      So it is  ✅ Safest mode and you have
                                ✅ Full control over every edit
                        
                        
	Edit automatically : ( here u are giving all the permission to claude to access your current Workspace/Project ) 
                           So Claude can directly edit files without asking for approval each time..
                         
                         
	Plan Mode : ( If u are doing multi-step operation & every step is very very important , always go with Plan mode )
                  Claude shows a plan before coding !
                  What it does : Claude analyzes the task and creates a plan before making changes.

            Simple explanation:
            Before touching any code, Claude says:

            Step 1 → Analyze project
            Step 2 → Modify API layer
            Step 3 → Update tests
            Step 4 → Verify everything

            Only after showing the plan does it proceed.
		    ( So first Claude present you entire plan , then return you code ) 
			Importance of this : YOur tokens will not waste !
            
            
            
    Auto Mode : Claude automatically chooses the best mode for the task.

        Simple explanation:
            Claude decides:

            Is there is a need of Small change?
            → CLAUDE will Edit Automatically

            Is there is a need of Complex change?
            → CLAUDE will go with Plan Mode

            Is there is a need of Risky operation?
            → CLAUDE will go with Ask Before Edit
            
            
            
            
       🎚️ Effort Level Slider : Controls how deeply Claude thinks
           Low Effort
               Claude will Think quickly
               Claude will Respond quickly

            Good for:
                Simple questions
                Small code fixes
                
                
                
          Medium Effort
                When u want Balanced reasoning
                When u want Balanced speed

          Good for:
                Most development work
            
            
            
            
          High Effort
                When u want More analysis
                When u want More planning
                ❌ Uses more tokens and takes longer



	right side of Ask before edit : Context 
	
	If u are working on any file & if u launch Claude code , this file will automatically got selected as Context !






'''
```
