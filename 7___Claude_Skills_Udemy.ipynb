# 2.. What are Skills : Real - World Scenario ?


````python
'''

Hi there 👋 , In the last session …
We understood WHY Skills are important !

Now lets understand : 👉 WHAT exactly are Skills ?


🎯 Definition :

    Skills are a FOLDER inside a Project where we keep DIFFERENT files & 
    with the help of these files , Agents/LLMs are able to understand HOW to carry a SPECIALISED task !
    
    
    In simple words :
        👉 Claude will load Skills ONLY in context , WHEN IT IS NEEDED !


Example : 
    When you say to Claude — "Hey Claude , Extract Amazon Users data from this API"
    Then only , Claude will go to the Skill folder !
    load necessary files , follow instructions of that file & THEN extract data for you 🚀




Now lets understand the STRUCTURE of Skills 📂 :


    👉 Skills is nothing but a FOLDER inside a project !


    Here is the basic hierarchy :

    📁 Project folder
        └── 📁 .claude folder
              └── 📁 skills folder
                    └── 📁 skill folder
                          └── 📄 SKILL.md   ← (Very important file !)



👉 Inside the skill folder , we have a file called SKILL.md
    

👉 Inside SKILL.md file , ALL the detailed instructions have been written 
   to perform some specific task !

    eg : For example, imagine that every day you need to:

        Extract Updated data from an API
        Clean and process the data
        Generate a report & Submit the report to managers
    
    
Since this is a repeated workflow, creating a Skill provides:

        ✅ Reliability
        ✅ Consistency
        ✅ Specialized instructions that don't need to be re-written every time
    
    Instead of repeating instructions every day , So lets create Skill for this !
    In SKILL.MD , u can mention : 

        👉 Which API endpoint to call ?
        👉 Which col-names to extract from the response ?
        👉 How to Handle timeout and authentication errors ?
        👉 What about Output format of Data : CSV, Excel, JSON, or SQLite

        
    All these instructions live inside SKILL.md !



👉 Apart from this , if YOUR Skill needs RESOURCES to perform that task ,
   then you can actually create SUB-FOLDERS & put resources inside that !


    📁 skill folder
        ├── 📄 SKILL.md
        ├── 📁 scripts/        ← sub-folder
        ├── 📁 templates/      ← sub-folder
        └── 📁 resources/      ← sub-folder



Lets understand these sub-folders with examples :

    👉 scripts/ folder :
            Imagine u are extracting Amazon data from API & u have created some scripts ie 
            Data Extraction scripts , Data cleaning scripts , KPI calculations ,Dashboard generation & they are working fine ! ..

            So instead of asking Claude code to generate these scripts from scratch every time , 
            u can say , Hey Claude code :
            These are the all the scripts , go & run it 
    
            So it will actually save lot of tokens bcz now u your Claude CODE will not write code , 
            It will reuse the existing scripts !



            So u can actually put all the scripts inside scripts/ folder ! 🐍
            
            
            
        


    👉 In resources/ folder :
                
        Imagine u are extracing data & u have some secrets & keys , so u can attach that in resources folder :
        Let say, you are extracting data using Pyspark & you expect Claude code to write latest Pyspark Code 
        May be Claude will not aware about latest version of Pyspark so it may give u some outdated code !
    
        & thats where in resources folder , u can attach these things !
    
        Similarly documentation of doing special tasks !
        That's where u can use resources ! 


            ie , Any extra reference material , images , data files , etc.
            can go inside resources/ folder !





    👉 In templates/ folder :
            Like u are creating PPTs & for that u may need :
                👉 Company design guidelines 🎨
                👉 Slide layouts
                👉 Color palettes
            You can put these into templates/ folder ! ✨





🎯 Proper Hierarchy of a Project with Multiple Skills :

    📁 Project folder
        └── 📁 .claude folder
              └── 📁 skills folder
                    ├── 📁 skill folder 1
                    │     ├── 📄 SKILL.md
                    │     └── 📁 optional resources
                    ├── 📁 skill folder 2
                    │     └── 📄 SKILL.md           ← (this is another Skill !)
                    └── 📁 skill folder 3
                          └── 📄 SKILL.md           ← (this is another Skill !)





And in the next session …
We'll go EVEN DEEPER and understand the internal structure of SKILL.md file 
( YAML Front-matter + Markdown Body ) — this is where the REAL magic happens 🔥


'''
````

# 3.. Skills Deep-dive : YAML Front-matter + Markdown body


````python
'''

Hi there 👋 , In the last session …
We understood WHAT Skills are & the FOLDER structure of Skills !






So now , lets deep dive into how to create Skills !

[Show GITHUB datasets]
Imagine that every day you need to:

        Extract Updated data from an API
        Clean and process the data
        Generate a report & Submit the report to managers
    
    
Since this is a repeated workflow, creating a Skill makes sense !


    
Instead of repeating instructions every day , You can create Skill for this !

So before creating Skill , first lets understand components of Skill & then in the next session , 
I will show u practical Part !




Lets understand : 👉 What does the Skill file which is SKILL.md ACTUALLY contains ?


     👉 SKILL.md actually has 2 IMPORTANT things :

        1) YAML Front-matter
        2) Markdown Body


Lets understand each of these one by one 👇




═══════════════════════════════════════════════
1) 📝 YAML Front-matter
═══════════════════════════════════════════════

YAML Front-matter is the TOP section of the SKILL.md file !
It has 2 things :

    👉 a) Skill name
    👉 b) Description



👉 a) Skill name :
            Using skill-name , Claude can IDENTIFY the skill !

👉 b) Description :
            Description is VERY IMPORTANT bcz ..
            After reading the description , Claude knows WHEN to use or load this Skill !

    Example :-   "Use this Skill to Fetch data from APIs and handles responses. 
                  Use when interacting with external APIs or retrieving data from web services."  



    🔥 So , description is like a TRIGGER for Claude ,





═══════════════════════════════════════════════
2) 📄 Markdown Body
═══════════════════════════════════════════════

    Here we WRITE DOWN EVERYTHING :-
    
    👉 All the instructions
    👉 + Code (if needed)
    👉 + Coding patterns
    👉 + Validation patterns


    + Link All files/references , like if u need Python script for Data Extraction  ,
       so here you can LINK that !



🎯 Lets visualize how a SKILL.md file looks :


```md
┌──────────────────────────────────────────────────────────┐
│  ---                                                     │
│  name: fetch-amazon-data                                 │
│  description: Use this Skill to fetch the six Amazon     │
│               datasets (inventory, products, returns,    │
│               reviews, sellers, shipping) from the       │
│               GitHub resources repo and save them as     │
│               timestamped CSVs with a run log            │
│  ---                                                     │
│                                                          │
│  # 📦 Fetch Amazon Data from API                        │
│                                                          │
│  ## Instructions                                         │
│                                                          │
│                
│  - Fetch all six CSV URLs                │
│  - Save data to a timestamped data/ folder          │
│        (format: YYYY-MM-DD_HH-MM-SS)                     │
│  - Write a run log: URLs called, successes,              │
│        failures, and any errors encountered              │
│                                                          │
│  ## 📚 Files Used                                        │
│                                                          │
│  - scripts/fetch.py                                      │
│        → Async download + save logic                     │
│                                                          │
│  - data/<timestamp>/                                     │
│        → Saved CSV files (one per dataset)               │
│                                                          │
│  - logs/<timestamp>/fetch.log                            │
│        → Log of every API call made                      │
└──────────────────────────────────────────────────────────┘
```


    Once Skill is defined , Then test your skill !
    If it is working fine , then rely on skill !
        Else improve this skill
    So it's like a iterative loop 
    
    





🟢 Quick Summary :

    👉 YAML Front-matter  =  Identity card (name + description)
    👉 Markdown Body      =  The detailed instruction manual

    👉 Name        →  Tells Claude — "Who am I ?"
    👉 Description →  Tells Claude — "WHEN to use me ?"
    👉 Body        →  Tells Claude — "HOW to do the task ?"



& in the next session …
We will understand HOW Claude actually LOADS these Skills internally —
This is where you'll learn about a beautiful concept called : "Progressive Disclosure" 🚀


'''
````

# 5.. How Does Claude load Skills ?


````python
'''

Hi there 👋 , In the last session …
We understood the internal structure of SKILL.md ( YAML + Body ) & we have hands-on as well !





One important doubt which most of Leaners have : 
Why to use Skills when we have CLAUDE.md ?


Ques : Can't I put this skill instructions in CLAUDE.md file & it becomes context for the LLM !
       Then , Why to create separate folder & file ?




Even if I put all the instrcutions in CLAUDE.md  , then if i am not using my skill , then I am wasting my tokens !
Bcz CLAUDE.md always loaded !



But If I create skills , so SKILL Name only gets loaded in CLAUDE context , not entire skill itself !
& if prompt matches skills , then only entire skill loaded into context!

ie body of skill(SKILL.md) only go to claude code context when Claude code invokes this skills !










Now , lets understand : 👉 HOW does Claude actually LOAD Skills ?
This is one of the MOST elegant concepts in the entire Claude ecosystem 🚀


🎯 The Big Idea : Progressive Disclosure


👉 Claude follows the idea of : PROGRESSIVE DISCLOSURE

    What does this fancy word mean ?
    
        👉 i.e Don't present info until it is needed !
                            OR
        👉 i.e Don't load that info in context window until it is needed !



🤔 Why does Claude do this ?

    See , Context Window is LIMITED !
    That's why , CLAUDE don't load Skill in memory until it is loaded ..
    ( Bcz if we dump everything at once , context window will FILL UP very fast 🚨 )




🎯 Progressive Disclosure works at 3 LEVELS :


═══════════════════════════════════════════════
Level 1 :  The Description
═══════════════════════════════════════════════

    👉 The Skill's YAML front-matter ALWAYS gets loaded ,
       as it is very VERY small !

    👉 So if you have 10 Skills , then all the 10 Skill names & descriptions 
       actually got loaded in context !


    Why ?
        👉 Bcz Claude needs to KNOW : "What Skills are available to me ?"




═══════════════════════════════════════════════
Level 2 :  SKILL.md ( loaded on-demand )
═══════════════════════════════════════════════

    So , when user asks to "fetch Amazon-data from Github ",
    Then Claude will SEARCH whether Skill is present or not ..
    
    👉 If Skill is available , then Claude will load body of that Skill & follow instructions !


    🎯 This is the BEAUTY of Progressive Disclosure :
        ( ie : The moment it is actually needed , then only Claude will load/show it ! )






═══════════════════════════════════════════════
Level 3 :  Supporting files / Resources
═══════════════════════════════════════════════

    👉 Even WITHIN SKILL.md , Claude is smart !
    👉 If SKILL.md references other files ( like scripts/ , templates/ ) ,
       Claude will load THOSE files ONLY when actually needed !


    So the loading happens like this 🎯 :

        Step 1 :  YAML loads instantly ( tiny )
        Step 2 :  When user asks for that task → SKILL.md body loads
        Step 3 :  If body references a file → that file loads
        





🟢 Lets visualize this :


    User : "Fetch Amazon Data from Github"
          ↓
    [ Claude scans all loaded YAML descriptions ]
          ↓
    [ Finds : "Fetch-amazon-data" Skill matches ]
          ↓
    [ Loads SKILL.md body into context ]
          ↓
    [ SKILL.md says : "Use these GitHUB URLs to fetch data" ]
          ↓
    [ Fetch data from Github Repo ]



🟢 Summary :

    👉 Level 1 : YAML descriptions  →  Always loaded ( small , cheap )
    👉 Level 2 : SKILL.md body      →  Loaded when needed
    👉 Level 3 : Supporting files   →  Loaded when referenced

And in the next session …
We'll explore the SKILLS MARKETPLACE — 
where you can DOWNLOAD ready-made Skills from the community or create your own !


'''
````
