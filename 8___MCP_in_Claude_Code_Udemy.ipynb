```python
'''

Hi there 👋 , Welcome to one of the MOST IMPORTANT sessions in this entire course ! 🚨

In the last 3 sessions , we got EXCITED about MCP — & how it gives Claude SUPER POWERS !
But with great power , there comes great responsibility 🕷️ ..


🚨 So once you give Claude access to your Browser , your GitHub , your DB —
    
    You MUST know :
        - What Claude can DO 
        - What Claude can SEE
        - What you should NEVER allow
        - How to stay SAFE
        
        & thats where Permissions & Tools comes into picture !





═══════════════════════════════════════════════
🎯 Section 1 : Understanding In-built Tools + MCP Tools 🛠️
═══════════════════════════════════════════════

First Understand what In-built Tools are , then we will understand with respect to External Tools (Browser , GitHub , your DB )



📌 What are Tools?

    Claude Code internally uses different tools to complete tasks.

    Examples:

    Reading files → uses Read Tool
    Editing code → uses Edit Tool
    Running terminal commands → uses Bash Tool
    Searching internet → uses WebSearch Tool

    👉 Claude automatically understands:

    which tool to use
    when to use it

    You don’t manually control every tool.





📌 What are Permissions?

    Permissions mean:

    You approve Claude to use certain tools or commands.
                OR
    Claude Code automatically uses built-in or MCP-provided tools or prompts without your approval ..


    Example:

    Allow Claude to use Edit Tool ( it means now Claude will not ask u to give Permissions when it will run Edit Tool  )
    Allow Claude to run Git commands
    Allow Claude to access WebSearch
    Allow GITHUB MCP to access repo , Projects from Github
    Allow playwright MCP to add Products in cart , but u will not give payment rights to CLAUDE !

    



📌 Where Permissions are Stored ?

    Permissions are stored inside : .claude/settings.local.json
    This file contains:

    allowed built-in tools/MCP Tools ( webfetch , read_sql)
    allowed commands
    deny rules





Open VS Code : 
    Either u can directly create settings.local.json file at Project Root & Mention permissions for the tools !
                    Or 
    You can also access Claude Code via Terminal as well !
        /permissions -> Allow -> Add a new rule -> Search "websearch"

    Click on "Add a new Rule"
    Search as "websearch" Tool





  Where should this rule be saved?
  ❯ 1. Project settings (local)  Saved in .claude\settings.local.json
    2. Project settings          Checked in at .claude\settings.json
    3. User settings             Saved in at ~/.claude/settings.json
    
    
    Meaning of Each Option
    1. Project settings (local)
        Applies only to the current project on your system.
        
    2. Project settings
        Shared with everyone accessing the project via the GitHub repository.
    
    3. User settings
        Applies globally to all Claude Code projects on your system



When u hit enter , u can see file as : "settings.local.json"

In settings.local.json

{
  "permissions": {
    "allow": [
      "websearch"
    ]
  }
}


## u can add "read" & "write" tool as well !

{
  "permissions": {
    "allow": [
      "websearch",
      "read",
      "write"
    ]
  }
}



Now if u want to cross check,  give prompt as :
    Create test.txt file

    Now Claude will not ask permission to you to create this File !


    ✅ Allowed
    Claude can:
        Read project files
        Edit files
        Use WebSearch












{
  "enabledMcpjsonServers": [
    "ddg-search",
    "playwright",
    "sqlite"
  ],
  "disabledMcpjsonServers": [
    "playwright_own",
    "tavily"
  ]



->> enabledMcpjsonServers / disabledMcpjsonServers

    These are not permissions.
    They simply control which MCP servers Claude Code loads from your MCP configuration.

    
    For example:

    "enabledMcpjsonServers": [
      "sqlite",
      "playwright"
    ]

    means : Load these MCP servers when Claude Code starts.




    Whereas :: 

    "disabledMcpjsonServers": [
      "tavily"
    ]

    means : Ignore this MCP server, even if it's defined in your MCP configuration.



'''
```
