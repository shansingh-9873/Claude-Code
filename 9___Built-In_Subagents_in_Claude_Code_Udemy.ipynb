# 1.. What are Sub-Agents ? 

```python
'''

                  🟦  (Parent)Main Agent / Claude Code  🟦
                                 |
              -------------------+-------------------
              |                  |                  |
        Sub-agent 1         Sub-agent 2        Sub-agent 3
        (Data Collection)    (Cleaner)          (Analyser)
        (own context win)   (own context win)  (own context win)



These sub-agents can't talk with each other  , bcz they have their separate context window !
They only talk with Parent agent ( ie Claude Code )


👉 Subagents are AI Assistants that run inside their OWN
   ISOLATED Context Window ..

   Sub-agent  =  Mini AI agent  +  its own SEPARATE context window








Now the que is : Why there is a need of Sub-agent ? 
Lets understand using an simple example :





------------------------------------------------------------
🔹 First , lets understand : How does an LLM actually work ?
------------------------------------------------------------

👉 See we know , LLMs are STATELESS / MEMORY-LESS .
   It means , They DO NOT have memory of past conversations on their own !


Lets see an example :

Call 1 :
   You ask  ->> "What is capital of India ?"
   LLM     ->> "Delhi"   ✅


Call 2 :
   You ask  ->> "What about Germany ?"
   LLM     ->> "What about Germany in what context ?" 🤯

   Why ?
   Because LLMs don’t have memory ..
   They don’t know what you asked in Call 1 !


------------------------------------------------------------
🔹 So how did Claude developers fix this ?
------------------------------------------------------------

They found a solution :
👉 When you ask a question to LLM ,
   we don't just send the new question -
   we send  ->>   new query + entire previous conversation

So now LLM will have full context of what was discussed before ✅

This is the trick that gives ChatGPT , Claude , Gemini their "memory feel" !


            new query 
                +                   ----->  LLM   (now has full context)
        entire previous conversation






'''
```

```python
'''

Let Understand it using a Real World Example : 

------------------------------------------------------------
🔹 Real Example  ->>  Imagine you are a Data Analyst @ Amazon 
------------------------------------------------------------

You are using Claude Code to do your day to day work ..


TURN 1 :
   You say  ->> "Hey CLAUDE, collect Amazon data for me from this GITHUB repo.."
   Lets say  ->>  this turn consumed roughly ≈ 30K tokens

   ie till Turn 1  ≈  30K tokens spent ✅


TURN 2 :
   Now you say  ->> "Clean this data .."

   But remember -  Claude has to send :
        👉  query  ( Clean this data )
                +
        👉  previous Turn 1 conversation  ( 30K tokens )
              
   Total sent to LLM  ≈  30K tokens
   LLM does the cleaning & returns new cleaned output  ( imagine 5K tokens )

   ie till Turn 2  ≈  (30K in 1st Turn + 5K in 2nd Turn) ie 35K tokens have been spent .. 😳


TURN 3 :
   Now you ask LLM  ->> "Perform Data Analysis on it"

   Again Claude sends  ->>  query + previous conversation [ 35K ]
   LLM responds with EDA output  ( imagine ≈ 20K tokens )

   ie till Turn 3  ≈  35K tokens  +  20K (EDA O/P)
                   =  55K tokens spent !! 🤯


Turn 4 : 
    Generate Reports !




You can clearly see ..
👉 Your context window is getting FILLED so fast ..
👉 And each call is costing you MORE money than the last one ..



------------------------------------------------------------
🔹 So there is a MAJOR problems with this approach
------------------------------------------------------------

PROBLEM 1 :  Context Window Overflow
   👉 Every model has a fixed context window ( like 200K , 1M etc )
   👉 If you keep stacking conversation , it fills very fast ..
   👉 Once full -> LLM forgets or breaks ..




------------------------------------------------------------
🔹 So whats the solution ?
------------------------------------------------------------

👉 This is EXACTLY why the idea of  ->>  SUBAGENTS came into picture 🚀

Subagents solve this problems in one shot by creating multiple Agents !




                  🟦  (Parent)Main Agent / Claude Code  🟦
                                 |
              -------------------+-------------------
              |                  |                  |
        Sub-agent 1         Sub-agent 2        Sub-agent 3
        (Data Collector)    (Cleaner)          (Analyser)
        (own context win)   (own context win)  (own context win)




------------------------------------------------------------
🔹 What is the advantage of this ?
------------------------------------------------------------

Q : What is adv. of this ?

Ans : 
   👉 Your context window will NOT fill so fast
   👉 Which automatically results in  ->> 
     👉  LOWER COST 💰 + 👉 better quality responses + 👉 cleaner separation of work


Thats actually the MAGIC of subagents ..

These sub-agents can't talk with each other  , bcz they have their separate context window !
They only talk with Parent agent ( ie Claude Code )





'''
```

# 2.. Built-in Subagents: Explore, Plan & General-Purpose


```python
'''

Hi there , in the last session , we have learnt what the sub-agents are !
Now lets first talk about types of sub-agents !



So there are 2 types of Sub-agents !


      [diagram]
                            🤖  Sub-agents
                                  |
                  ----------------+----------------
                  |                                |
          Built-in Sub-agents              Custom Sub-agents
          (given by Claude)                (built by YOU)
          - Explore                         - sales-data-analyst
          - Plan                            - security-auditor
          - General Purpose                 - blog-post-writer
                                            - jira-ticket-writer
                                            - ... anything YOU want



   👉 Explore SUb-Agent          ->> To understand codebase  /  summarise repo
           [ show VS CODE Project]
       like u have 10 Data sources ( csv , db , Big data  , Json etc) in your Project !
            10 CSV files
            5 Databases
            3 JSON files
            2 Data Lakes
            
            
       Normally, you might ask the parent Claude Code session:
        "Explore all the Data Sources.."
            The problem?
        The parent session has to explore every data source itself, consuming its own context window..
        
        
       So instead of asking CLAUDE CODE Parent session to understand these sources  , 
       u can launch one EXPLORE SUB-AGENT per data source ( 1 for each source) ,
       & these 4 sub-agent will give summary about all those 10 sources !
       
       so these Sub-agents will run parallelly & each agent will have its own Context Window !
       
       
       
       So if u have large, independent tasks (multiple folders, repositories, files, services, etc.), 
       then using sub-agents often reduces the parent's workload and can improve overall efficiency
       
       
       
       
   
   👉 Plan SUb-Agent (in Plan mode)       
       The Plan sub-agent is designed for planning before implementation..
       
           User : Generate an implementation plan for Dashboard using Python, SQL & Streamlit
                   Instead of immediately writing code, Claude launches a Plan Sub-agent.
                   The Plan Sub-agent will:
           
           • Researches the codebase
           • Gathers context
           • Produces step-by-step implementation plans
   
   
   
   👉 General-Purpose SUb-Agent  ->>  
       The General-Purpose sub-agent is the most flexible built-in sub-agent.
        It doesn't specialize in just exploring or planning.
        Instead, it can handle a wide variety of software engineering tasks..
        
           If u are doing : 
               Read + Write
               Research + implementation
               Code modifications
               🐞 Debug issues
               🔍 Research libraries









Every agent have :
   ✅ its own SYSTEM PROMPT
   ✅ its own set of SKILLS
   ✅ its own set of TOOLS  ( only what that agent actually needs )
   ✅ DIFFERENT MODEL access

Example :

   Agent 1   ->>   Opus    ( heavy reasoning  -> use most powerful model )
   Agent 2   ->>   Haiku   ( simple lookup tasks -> use cheaper / faster model )
   Agent 3   ->>   Sonnet  ( balanced tasks  -> mid-tier model )










'''
```
