# 4.. How Claude Code works Internally ?

```python
'''

See we know : 
Claude Code = LLM + Built-in tools 


Although u can add custom tools using MCP , which we will learn in near future , but just for sake of simplicity
lets understand how does CLAUDE CODE works internally  , the moment u give some task or any Prompt !






Query ->> Read this Database file !
User Query
    ↓
"Read this database file"
    ↓
LLM
    ↓
Generates Python code
    ↓
At this stage:
Pure Generative AI
(Because the model is only predicting text/code)


It is only generating text such as:
import sqlite3
import pandas as pd

conn = sqlite3.connect("sales.db")
df = pd.read_sql("SELECT * FROM Users", conn)    
    
    
    
    
    
It means , LLMs are doing their work , but they cannot read files, create files, 
        run code, or access your computer.

but when Tools comes , Agentic Nature comes !


Tools[ Receives code as I/P ] ->> create new file ( using Write Tool) in our codebase 
        How Write tools create new file in our codebase ?
            A Write Tool might internally use some command like : 
                with open("explore_db.py", "w") as f:
                    f.write(code)







User Query
    ↓
"Read this database file"
    ↓
LLM
    ↓
Generates Python code
    ↓
At this stage: Pure Generative AI
(LLM is only generating text/code)
But LLM cannot read files, create files, run code, or access your computer.
    ↓
Generated code is passed to a Write Tool
    ↓
Write Tool Creates a new Python file in the codebase
    ↓
Tool internally uses File System APIs
(e.g., open(), write(), Path.write_text())
    ↓
New file successfully created
    ↓
Agent observes the result
    ↓
Task completed



'''
```
